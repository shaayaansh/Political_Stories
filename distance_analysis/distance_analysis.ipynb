```python
import re
import nltk
from nltk.corpus import stopwords

# Download stopwords if not already downloaded
try:
    stopwords.words('english')
except LookupError:
    nltk.download('stopwords')

# Get English stopwords
STOPWORDS_SET = set(stopwords.words('english'))

print("NLTK stopwords loaded successfully.")
```

```python
# Function to tokenize text and remove stopwords
def tokenize_and_clean(text):
    # Convert to lowercase
    text = text.lower()
    # Find all words using regex
    words = re.findall(r'\b\w+\b', text)
    # Remove stopwords and return as a set for efficient lookup
    return set(word for word in words if word not in STOPWORDS_SET)

# Function to calculate distance based on shared words
# The closer to 1, the further the distance????
def calculate_distance(article_words, reference_words):
    if not article_words:
        return 1.0  # If article has no words, it's completely dissimilar

    # Count words from the article that are present in the reference
    shared_words_count = len(article_words.intersection(reference_words))

    # Calculate percentage of article words present in reference
    percentage_present = shared_words_count / len(article_words)

    # Distance is 1 - percentage
    return 1.0 - percentage_present


def return_similiar_words(article_words, reference_words):
    return_list = []
    if not article_words:
        return return_list  # If article has no words, it's completely dissimilar
    for word in article_words:
        if word in reference_words:
            return_list.append(word)
    return return_list

print("Helper functions `tokenize_and_clean` and `calculate_distance` defined.")
```

### Paired t-test for AI Climate Change Stories

A paired t-test is used when you want to compare the means of two related samples. In this case, for each AI climate change story, you have two measurements: its distance to pro-reference arguments and its distance to con-reference arguments. This makes them 'paired' observations.

**Assumptions for a Paired t-test:**
1.  **Paired Observations:** Each observation in one sample is matched with an observation in the other sample.
2.  **Independence:** The paired observations are independent of each other.
3.  **Normality:** The differences between the paired observations are approximately normally distributed. (The t-test is robust to deviations from normality for larger sample sizes).
4.  **No Outliers:** There are no extreme outliers in the differences between the paired observations.

Let's apply this to the AI climate change stories.

```python
from scipy import stats

# Filter for AI climate change stories
ai_climate_df = df[(df['author'] == 'ai') & (df['topic'] == 'climate change')]

# Extract the distances
distance_pro = ai_climate_df['distance_to_pro_ref']
distance_con = ai_climate_df['distance_to_con_ref']

# Perform the paired t-test
t_statistic, p_value = stats.ttest_rel(distance_pro, distance_con)

print(f"T-statistic: {t_statistic:.4f}")
print(f"P-value: {p_value:.4f}")

# Interpretation
alpha = 0.05  # Significance level

if p_value < alpha:
    print(f"\nSince the p-value ({p_value:.4f}) is less than the significance level ({alpha}), we reject the null hypothesis.")
    print("There is a statistically significant difference between the mean distance to pro-reference arguments and the mean distance to con-reference arguments for AI climate change stories.")
else:
    print(f"\nSince the p-value ({p_value:.4f}) is greater than the significance level ({alpha}), we fail to reject the null hypothesis.")
    print("There is no statistically significant difference between the mean distance to pro-reference arguments and the mean distance to con-reference arguments for AI climate change stories.")

# Additionally, let's look at the mean differences to understand the direction
mean_distance_pro = distance_pro.mean()
mean_distance_con = distance_con.mean()

print(f"\nMean distance to pro-reference: {mean_distance_pro:.4f}")
print(f"Mean distance to con-reference: {mean_distance_con:.4f}")

if mean_distance_pro < mean_distance_con:
    print("On average, AI climate change stories are closer to pro-reference arguments.")
elif mean_distance_con < mean_distance_pro:
    print("On average, AI climate change stories are closer to con-reference arguments.")
else:
    print("On average, AI climate change stories have similar distances to both pro and con reference arguments.")
```

### Independent Samples t-test: AI vs. Human Climate Stories

Now, let's compare the 'distance_to_pro_ref' and 'distance_to_con_ref' between AI-generated climate change stories and human-generated climate change stories using an independent samples t-test.

```python
from scipy import stats

# Filter for AI climate change stories
ai_climate_df = df[(df['author'] == 'ai') & (df['topic'] == 'climate change')]

# Filter for Human climate change stories
human_climate_df = df[(df['author'] == 'human') & (df['topic'] == 'climate change')]

# --- Comparison for 'distance_to_pro_ref' ---
print("\n--- Comparing 'distance_to_pro_ref' (AI vs. Human) ---")

distance_pro_ai = ai_climate_df['distance_to_pro_ref']
distance_pro_human = human_climate_df['distance_to_pro_ref']

t_statistic_pro, p_value_pro = stats.ttest_ind(distance_pro_ai, distance_pro_human, equal_var=True) # Assuming equal variances for now

print(f"T-statistic (Pro Ref): {t_statistic_pro:.4f}")
print(f"P-value (Pro Ref): {p_value_pro:.4f}")

alpha = 0.05
if p_value_pro < alpha:
    print(f"  Reject H0: Statistically significant difference in 'distance_to_pro_ref' between AI and Human climate stories.")
else:
    print(f"  Fail to reject H0: No statistically significant difference in 'distance_to_pro_ref' between AI and Human climate stories.")

print(f"  Mean 'distance_to_pro_ref' (AI): {distance_pro_ai.mean():.4f}")
print(f"  Mean 'distance_to_pro_ref' (Human): {distance_pro_human.mean():.4f}")


# --- Comparison for 'distance_to_con_ref' ---
print("\n--- Comparing 'distance_to_con_ref' (AI vs. Human) ---")

distance_con_ai = ai_climate_df['distance_to_con_ref']
distance_con_human = human_climate_df['distance_to_con_ref']

t_statistic_con, p_value_con = stats.ttest_ind(distance_con_ai, distance_con_human, equal_var=True) # Assuming equal variances for now

print(f"T-statistic (Con Ref): {t_statistic_con:.4f}")
print(f"P-value (Con Ref): {p_value_con:.4f}")

if p_value_con < alpha:
    print(f"  Reject H0: Statistically significant difference in 'distance_to_con_ref' between AI and Human climate stories.")
else:
    print(f"  Fail to reject H0: No statistically significant difference in 'distance_to_con_ref' between AI and Human climate stories.")

print(f"  Mean 'distance_to_con_ref' (AI): {distance_con_ai.mean():.4f}")
print(f"  Mean 'distance_to_con_ref' (Human): {distance_con_human.mean():.4f}")
```

### Independent Samples t-test for All Topics (AI vs. Human)

Let's generalize the comparison and run the independent samples t-test for all topics present in our DataFrame, comparing AI-generated stories against human-generated stories for both pro and con reference distances.

```python
from scipy import stats

# Get all unique topics
all_topics = df['topic'].unique()

print("--- Performing Independent Samples t-tests for all topics (AI vs. Human) ---")

for topic in all_topics:
    print(f"\n\nTopic: {topic.upper()}")
    print("-------------------------------------------------------------------")

    # Filter for the current topic
    topic_df = df[df['topic'] == topic]

    # Separate AI and Human stories for the current topic
    ai_topic_df = topic_df[topic_df['author'] == 'ai']
    human_topic_df = topic_df[topic_df['author'] == 'human']

    # Check if there's enough data for both groups for the current topic
    if len(ai_topic_df) == 0 or len(human_topic_df) == 0:
        print(f"  Skipping {topic} due to insufficient data for AI or Human stories.")
        continue

    # --- Comparison for 'distance_to_pro_ref' ---
    print("\n  --- Comparing 'distance_to_pro_ref' (AI vs. Human) ---")

    distance_pro_ai = ai_topic_df['distance_to_pro_ref']
    distance_pro_human = human_topic_df['distance_to_pro_ref']

    # Perform t-test only if both samples have data
    if len(distance_pro_ai) > 1 and len(distance_pro_human) > 1:
        t_statistic_pro, p_value_pro = stats.ttest_ind(distance_pro_ai, distance_pro_human, equal_var=True)

        print(f"  T-statistic (Pro Ref): {t_statistic_pro:.4f}")
        print(f"  P-value (Pro Ref): {p_value_pro:.4f}")

        alpha = 0.05
        if p_value_pro < alpha:
            print(f"    Reject H0: Statistically significant difference in 'distance_to_pro_ref'.")
        else:
            print(f"    Fail to reject H0: No statistically significant difference in 'distance_to_pro_ref'.")

        print(f"    Mean 'distance_to_pro_ref' (AI): {distance_pro_ai.mean():.4f}")
        print(f"    Mean 'distance_to_pro_ref' (Human): {distance_pro_human.mean():.4f}")
    else:
        print("    Not enough data in both AI and Human groups for 'distance_to_pro_ref' comparison.")

    # --- Comparison for 'distance_to_con_ref' ---
    print("\n  --- Comparing 'distance_to_con_ref' (AI vs. Human) ---")

    distance_con_ai = ai_topic_df['distance_to_con_ref']
    distance_con_human = human_topic_df['distance_to_con_ref']

    # Perform t-test only if both samples have data
    if len(distance_con_ai) > 1 and len(distance_con_human) > 1:
        t_statistic_con, p_value_con = stats.ttest_ind(distance_con_ai, distance_con_human, equal_var=True)

        print(f"  T-statistic (Con Ref): {t_statistic_con:.4f}")
        print(f"  P-value (Con Ref): {p_value_con:.4f}")

        if p_value_con < alpha:
            print(f"    Reject H0: Statistically significant difference in 'distance_to_con_ref'.")
        else:
            print(f"    Fail to reject H0: No statistically significant difference in 'distance_to_con_ref'.")

        print(f"    Mean 'distance_to_con_ref' (AI): {distance_con_ai.mean():.4f}")
        print(f"    Mean 'distance_to_con_ref' (Human): {distance_con_human.mean():.4f}")
    else:
        print("    Not enough data in both AI and Human groups for 'distance_to_con_ref' comparison.")
```
